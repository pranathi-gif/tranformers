```python


%pip install -q transformers torch pandas scikit-learn

import pandas as pd
from transformers import pipeline
from sklearn.metrics import accuracy_score

df = pd.read_csv(
    "/content/Reviews.csv",
    encoding="utf-8",
    on_bad_lines="skip")

print("Dataset Shape:", df.shape)
print(df.head())

df = df[["Text", "Score"]].copy()

df.columns = ["text", "score"]

print(df.head())




```

```python
df = df.dropna(subset=["text", "score"])

df = df[df["score"].isin([1, 2, 4, 5])].copy()

df["true_label"] = df["score"].apply(
    lambda x: "POSITIVE" if x >= 4 else "NEGATIVE"
)

print("\nSentiment Distribution:")
print(df["true_label"].value_counts())
```

```python
sample_df = df.sample(
    n=150,
    random_state=42
).copy()

sample_df = sample_df.reset_index(drop=True)

print("\nSample Shape:", sample_df.shape)
```

```python

sentiment_model = pipeline("sentiment-analysis")


predictions = sentiment_model(
    sample_df["text"].tolist(),
    truncation=True,
    max_length=512,
    batch_size=16
)
```

```python
sample_df["predicted_label"] = [
    result["label"].upper()
    for result in predictions
]

sample_df["confidence"] = [
    result["score"]
    for result in predictions
]
```

```python


print(f"Sample Size: {len(sample_df)}")

# Check required columns
print("Available Columns:", sample_df.columns.tolist())

# Calculate accuracy
accuracy = accuracy_score(
    sample_df["true_label"],
    sample_df["predicted_label"]
)

print(f"Test Accuracy: {accuracy * 100:.2f}%")
```

```python
print("\n========== EXAMPLE PREDICTIONS ==========")

for i, row in sample_df.head(5).iterrows():
    print(f"\nExample {i + 1}")
    print("Text:", row["text"][:300])
    print("True Label:", row["true_label"])
    print("Predicted Label:", row["predicted_label"])
    print("Confidence:", round(row["confidence"], 4))


display(
    sample_df[
        ["text", "true_label", "predicted_label", "confidence"]
    ].head(5)
)
```
